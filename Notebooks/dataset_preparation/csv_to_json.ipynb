```python
import pandas as pd
import numpy as np
import json
```

```python
path_to_flytech = "path/to/flytech.csv" #add your path to flytech.csv created using ....
path_to_leetcode = "path/to/leetcode.csv" #add your path to leetcode.csv created using leet
```

```python
flytech_df=pd.read_csv("./flytech.csv")
leetcode_df=pd.read_csv("./leetcode.csv")
taco_df=pd.read_csv("./taco_cleaned.csv")

flytech_df=flytech_df[['instruction','output']]

flytech_df=flytech_df.values
leetcode_df=leetcode_df.values[200:]
taco_df=taco_df.values

df4=np.vstack((flytech_df,leetcode_df))
final_df=np.vstack((df4,taco_df))
```

```python
json_f={}
index=1
for i in final_df:
  x={
      "instruction":i[0],
      "code":i[1]
  }
  json_f[f"question_{index}"]=x

  index+=1

with open("csv_to_json.json","w") as f:
  json.dump(json_f,f,indent=4)

with open("csv_to_json.json","r") as f:
  data=json.load(f)

with open("csv_to_jsonl.jsonl","w") as f:
  for key,value in data.items():
    f.write(json.dumps(value) + "\n")
```
